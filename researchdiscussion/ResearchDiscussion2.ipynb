### Presentation Summary

I really enjoyed Christopher Johnson’s presentation and what I found really interesting was how Spotify uses collaborative filtering and matrix factorization to recommend music. Unlike platforms like Netflix that rely on explicit ratings from users, Spotify works primarily with implicit feedback. So, instead of asking users to rate songs, it infers preferences from listening behavior, like whether a song was streamed and how frequently it was played. The system also assigns greater confidence to songs that are played repeatedly, allowing Spotify to distinguish between a casual listen and a strong preference. This can really help to uncover hidden patterns between users and artists and supports more personalized recommendations.

What stood out to me even more was that the biggest challenge wasn’t necessarily the recommendation algorithm itself; it was scaling it to handle millions of users and massive amounts of listening data. Johnson explained how Spotify moved from Hadoop to Spark because Spark could keep data in memory instead of constantly reading and writing to disk, making ALS much faster and more efficient.

I also thought it was interesting that even with Spark, they still ran into issues with memory usage, data shuffling, and failed executors when working with very large datasets. It was a good reminder that building a recommender system in theory is very different from running one at the scale of a company like Spotify. 

The presentation was relatable, and I was able to realize that I experience these recommendations every day without really thinking about how they are generated. Spotify often introduces me to artists I have never listened to before, but many of them end up matching my music taste surprisingly well. After watching this presentation, it makes more sense how Spotify can make those recommendations by looking at listening patterns across millions of users and assigning more weight to artists and songs that I repeatedly stream. It was interesting to see the machine learning and engineering infrastructure behind something that feels so simple from my perspective as a user.

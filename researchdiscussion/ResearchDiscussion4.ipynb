### Mitigating the Harm of Recommender Systems

One of the biggest challenges with recommender systems is that they are often designed to maximize engagement, not necessarily to promote accurate information or healthy user experiences. The more time people spend clicking, watching videos, or interacting with content, the more successful the system is considered. The problem is that content that makes people angry or sparks strong reactions often gets a lot of engagement (think ragebait). Since recommender systems are trying to keep users clicking, they can sometimes end up recommending more controversial or extreme content because that's what gets attention. In the Wired article, DiResta explains that once a user interacts with a certain type of content, recommendation algorithms often continue suggesting more of the same, which can gradually reinforce particular viewpoints or interests.

I think one way to reduce this problem is for platforms to look beyond simple engagement metrics. Instead of only asking "Will this keep someone clicking?", recommender systems should also consider factors like credibility, content quality, and potential harm. I also liked the idea of giving users more control over their recommendations. Sometimes it feels like the algorithm decides what we see without us really understanding why, so having more transparency and customization could help.

The article on social influence bias was interesting because it focused on a different type of problem. The researchers found that when people see average ratings before giving their own opinion, they often shift their ratings closer to the group consensus. In their study, ratings became significantly closer to the median after users saw what others had rated.

This made me think about sites like Amazon, Yelp, or even Rotten Tomatoes. If I see that thousands of people gave something a very high rating before I try it myself, it's hard not to let that influence my opinion. One simple way to reduce this bias would be to hide ratings until after a user submits their own review. That would encourage people to form their own opinions first instead of being influenced by the crowd.

Overall, I think recommender systems can be incredibly useful, but they also have a lot of influence over what information we consume and how we make decisions. As these systems become more common, companies need to balance engagement with responsibility and think more carefully about the unintended effects their algorithms can have on users.
